```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

# 设置中文字体
plt.rcParams['font.sans-serif'] = ['SimHei']  # 使用黑体
plt.rcParams['axes.unicode_minus'] = False  # 正确显示负号

# 创建图形和轴
fig, ax = plt.subplots(figsize=(12, 8))

# 实验观测值及其误差
experimental_value = 5.0
experimental_sigma = 0.5

# 三个模型的预测值及其误差
models = [
    {"name": "Model1 (low precision, high accuracy)", "median": 5.1, "sigma": 1.0, "x_pos": 1},
    {"name": "Model2 (low precision, low accuracy)", "median": 7.5, "sigma": 1.0, "x_pos": 2},
    {"name": "Model3 (high precision, high accuracy)", "median": 5.4, "sigma": 0.3, "x_pos": 3}
]

# 绘制实验观测值及其误差范围
ax.axhline(y=experimental_value, color='red', linestyle='-', linewidth=2, label='Experimental value')
ax.axhspan(experimental_value - experimental_sigma, experimental_value + experimental_sigma, 
           alpha=0.2, color='red')

# ... 其余代码保持不变 ...

# 设置图例
ax.legend(loc='upper right', fontsize=10)


# 绘制每个模型的误差棒和中位数
for i, model in enumerate(models):
    # 绘制误差棒
    ax.errorbar(model["x_pos"], model["median"], yerr=model["sigma"], 
                fmt='s', capsize=8, capthick=2, elinewidth=2,
                color='blue', markersize=8)
    
    # 在模型旁边添加描述文字
    descriptions = [
        "accurate and imprecise",
        "inaccurate and imprecise", 
        "accurate and precise"
    ]
    text_y = model["median"] + model["sigma"] + 0.15  # 文字位置=中位数+sigma+0.5
    ax.text(model["x_pos"], text_y, descriptions[i],
            ha='center', va='bottom', fontsize=12, weight='bold')
# 移除原有的precision和accuracy说明文字
# 设置图形属性
ax.set_xlim(0.5, 3.5)
ax.set_ylim(2, 9)
ax.set_xlabel('')  # 清空x轴标签
ax.set_ylabel('observable', fontsize=12)

ax.legend(loc='upper right', fontsize=10)
# 消除坐标轴上的数字
ax.set_xticks([1, 2, 3])
ax.set_xticklabels(['Model1', 'Model2', 'Model3'], fontsize=12, weight='bold')

ax.set_yticks([])  # 消除y轴数字
# 添加网格
ax.grid(True, linestyle='--', alpha=0.7)

plt.tight_layout()
plt.savefig('schematic.pdf')
plt.show()
```
