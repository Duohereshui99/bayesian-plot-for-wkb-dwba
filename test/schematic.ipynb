```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

# 设置中文字体
plt.rcParams['font.sans-serif'] = ['SimHei']  # 使用黑体
plt.rcParams['axes.unicode_minus'] = False  # 正确显示负号

# 创建图形和轴
fig, ax = plt.subplots(figsize=(12, 8))

# 实验观测值及其误差
experimental_value = 5.0
experimental_sigma = 0.5

# 三个模型的预测值及其误差
models = [
    {"name": "模型1 (低精度, 高准确度)", "median": 5.1, "sigma": 1.2, "x_pos": 1},
    {"name": "模型2 (低精度, 低准确度)", "median": 7.5, "sigma": 1.0, "x_pos": 2},
    {"name": "模型3 (高精度, 高准确度)", "median": 5.4, "sigma": 0.3, "x_pos": 3}
]

# 绘制实验观测值及其误差范围
ax.axhline(y=experimental_value, color='black', linestyle='-', linewidth=2, label=f'实验观测值: {experimental_value}')
ax.axhspan(experimental_value - experimental_sigma, experimental_value + experimental_sigma, 
           alpha=0.2, color='gray', label=f'实验值误差范围 (±{experimental_sigma}σ)')

# 绘制每个模型的误差棒和中位数
for model in models:
    # 绘制误差棒
    ax.errorbar(model["x_pos"], model["median"], yerr=model["sigma"], 
                fmt='o', capsize=8, capthick=2, elinewidth=2, 
                label=model["name"])
    
    # 添加模型名称
    ax.text(model["x_pos"], model["median"] - 2.2, model["name"], 
            ha='center', va='top', fontsize=10)

# 添加说明文本
ax.text(0.5, 8.5, 'Accuracy (准确度): 测量值与真实值的接近程度', fontsize=14, ha='center', weight='bold')
ax.text(0.5, 8.0, 'Precision (精确度): 多次测量值之间的接近程度', fontsize=14, ha='center', weight='bold')

# 设置图形属性
ax.set_xlim(0.5, 3.5)
ax.set_ylim(2, 9)
ax.set_xlabel('模型', fontsize=12)
ax.set_ylabel('观测值', fontsize=12)
ax.set_title('Precision与Accuracy概念示意图', fontsize=16, weight='bold')
ax.legend(loc='upper right', fontsize=10)

# 添加网格
ax.grid(True, linestyle='--', alpha=0.7)

plt.tight_layout()
plt.show()
plt.savefig('schematic.pdf')
```
