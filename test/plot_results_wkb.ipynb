```python
import numpy as np
import matplotlib.pyplot as plt


Sys=['211Po','213Rn','215Ra','217Th','219U']
Z=[84,86,88,90,92]
T_half_expt=[0.516,19.36e-3,1.68e-3,254e-6,60e-6]
T_half_calc=[0.5158905614040402,0.019376367726007926,0.0016799052057963173,0.0002540000639797968,5.960557474881356e-05]
T_half_2sigma=[0.005965097512233266,0.0003818913474164974,1.9873828913767867e-05,4.0074559194984095e-06,1.3631900306087726e-05]


# 计算比值和误差
ratios = [calc/expt for calc, expt in zip(T_half_calc, T_half_expt)]
ratio_errors = [sigma/expt for sigma, expt in zip(T_half_2sigma, T_half_expt)]

# 创建图形
plt.figure(figsize=(12, 6))

# 绘制比值点
for z, ratio, sys in zip(Z, ratios, Sys):
    plt.plot(z, ratio, 'o', markersize=5, label=f'{sys}')

# 绘制误差棒
plt.errorbar(Z, ratios, yerr=ratio_errors, fmt='none', 
             ecolor='black', elinewidth=2, capsize=5)

# 添加参考线
plt.axhline(y=1, color='gray', linestyle='--', linewidth=1)

# 设置坐标轴
plt.xticks(Z, [f"{z}({sys})" for z, sys in zip(Z, Sys)], fontsize=10)
plt.xlabel('Z', fontsize=12)
plt.ylabel('Theoretical/Experimental Ratio', fontsize=12)
plt.ylim(0.75,1.25)  # 自动调整y轴范围

# 添加图例和网格
plt.legend(fontsize=10, loc='upper left')
plt.grid(True, which="both", ls="--", alpha=0.4)

plt.tight_layout()
plt.savefig('plot_results_wkb.pdf')
plt.show()
```
